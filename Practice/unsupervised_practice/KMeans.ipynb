## K means clustering practice 

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.cluster import KMeans
import warnings

warnings.filterwarnings('ignore')

sns.set_theme(
    style="whitegrid",   
    context="talk",
    palette="pastel"       
)
```

# Trying it on iris dataset without labels

```python
from sklearn.datasets import load_iris

iris = load_iris()

X = iris.data

X
```

```python
from sklearn.preprocessing import StandardScaler

scalar = StandardScaler()

X_scaled = scalar.fit_transform(X)
```

```python
inertia = []
K_values = range(1,13)

for k in K_values:
    
    clusters = KMeans(n_clusters=k, random_state=42)
    
    clusters.fit(X_scaled)
    
    wcss = clusters.inertia_
    
    inertia.append(wcss)
    
```

```python
inertia
```

```python
plt.Figure(figsize=(12,8))
sns.scatterplot(x= K_values, y= inertia)
sns.lineplot(x= K_values, y= inertia)
```

```python
kmeans_3 = KMeans(n_clusters=3, random_state=42, n_init=10)
labels_3 = kmeans_3.fit_predict(X)

kmeans_4 = KMeans(n_clusters=4, random_state=42, n_init=10)
labels_4 = kmeans_4.fit_predict(X)
```

```python
print("K=3 inertia:", kmeans_3.inertia_)
print("K=4 inertia:", kmeans_4.inertia_)
```

```python
labels_3
```

```python
labels_4
```

```python
from sklearn.metrics import silhouette_score

score_3 = silhouette_score(X, labels_3)
score_4 = silhouette_score(X, labels_4)

print("K=3 Silhouette Score:", score_3)
print("K=4 Silhouette Score:", score_4)
```

```python
final_kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

labels = final_kmeans.fit_predict(X)
```

```python
print("Cluster Labels:")
print(labels)

print("\nCentroids:")
print(final_kmeans.cluster_centers_)
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    x=X[:, 0],  # Sepal Length
    y=X[:, 1],  # Sepal Width
    hue=labels,
    palette="viridis",
    s=70
)

plt.scatter(
    final_kmeans.cluster_centers_[:, 0],
    final_kmeans.cluster_centers_[:, 1],
    c="red",
    marker="X",
    s=200,
    label="Centroids"
)

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("K-Means Clustering, sepal size")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    x=X[:, 2],
    y=X[:, 3],
    hue=labels,
    palette="viridis",
    s=70
)

plt.scatter(
    final_kmeans.cluster_centers_[:, 2],
    final_kmeans.cluster_centers_[:, 3],
    c="red",
    marker="X",
    s=200,
    label="Centroids"
)

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("K-Means Clustering Petal size")
plt.show()
```

```python
y = iris.target
```

```python
print("Actual Labels:")
print(y)

print("\nK-Means Labels:")
print(labels)
```

```python
from sklearn.metrics import adjusted_rand_score

ari = adjusted_rand_score(y, labels)

print("Adjusted Rand Index:", ari)
```

```python
from sklearn.metrics import normalized_mutual_info_score

nmi = normalized_mutual_info_score(y, labels)

print("NMI:", nmi)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y, labels)

plt.figure(figsize=(12, 8))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Cluster 0", "Cluster 1", "Cluster 2"],
    yticklabels=["Setosa", "Versicolor", "Virginica"]
)

plt.xlabel("K-Means Cluster")
plt.ylabel("Actual Iris Class")
plt.title("Actual Iris Labels vs K-Means Clusters")

plt.show()
```

```python
from collections import Counter

for cluster in np.unique(labels):
    actual_classes = y[labels == cluster]

    counts = Counter(actual_classes)

    print(f"Cluster {cluster}:")
    print(counts)
```

```python
cluster_mapping = {
    0: 1,  # Cluster 0 → Versicolor
    1: 0,  # Cluster 1 → Setosa
    2: 2   # Cluster 2 → Virginica
}
```

```python
mapped_labels = np.array([
    cluster_mapping[label]
    for label in labels
])

print("Mapped Labels:")
print(mapped_labels)
```

```python
target_names = iris.target_names

mapped_species = np.array([
    target_names[label]
    for label in mapped_labels
])

print(mapped_species)
```

# let's predict a new flower 

```python
new_flower = [[5.1, 3.5, 1.4, 0.2]]
```

```python
new_flower_scaled = scalar.transform(new_flower)

pred_cluster = final_kmeans.predict(new_flower_scaled)[0]

pred_class = cluster_mapping[pred_cluster]

print("Predicted class:", pred_class)
print("Predicted species:", iris.target_names[pred_class])
```
