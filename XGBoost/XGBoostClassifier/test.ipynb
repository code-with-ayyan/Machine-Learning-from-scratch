```python
import numpy as np 
import pandas as pd 
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
from XGBoost import XGBoostClassifierCustom
import warnings 

warnings.filterwarnings('ignore')
```

```python
print("Hello World!")
print("My name is Ayyan")
print("How are you ?")
```

```python
X, y = make_classification(n_features= 2, n_informative=2, n_redundant=0,n_classes= 2, n_samples=1000, random_state=42)


X = pd.DataFrame(X, columns=['Feature1','Feature2'])

X
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

scalar = StandardScaler()

X_train = scalar.fit_transform(X_train)
X_test = scalar.transform(X_test)
```

```python
model = XGBoostClassifierCustom(n_estimators=50,
            learning_rate=0.3,
            max_depth=3,
            max_leaf_nodes=None,
            min_child_weight=1,
            gamma=0,
            reg_lambda=1,
            reg_alpha=0)


model.fit(X_train, y_train)
```

```python
model.n_estimators
```

```python
custom_pred = model.predict(X_test)
custom_pred
```

```python
from sklearn.metrics import accuracy_score

acc_custom = accuracy_score(y_test, custom_pred)

print("Accuracy : ",acc_custom)
```

```python
import plotly.graph_objects as go



mask_0 = y_test == 0
mask_1 = y_test == 1

x1_min = X_test[:, 0].min()
x1_max = X_test[:, 0].max()

x2_min = X_test[:, 1].min()
x2_max = X_test[:, 1].max()


x1 = np.linspace(
    x1_min,
    x1_max,
    50
)

x2 = np.linspace(
    x2_min,
    x2_max,
    50
)

X1, X2 = np.meshgrid(
    x1,
    x2
)




Z = np.full_like(
    X1,
    0.5
)




fig = go.Figure()




fig.add_trace(
    go.Scatter3d(
        x=X_test[mask_0, 0],
        y=X_test[mask_0, 1],
        z=y_test[mask_0],
        mode="markers",
        name="Actual Class 0",
        marker=dict(
            size=6
        )
    )
)




fig.add_trace(
    go.Scatter3d(
        x=X_test[mask_1, 0],
        y=X_test[mask_1, 1],
        z=y_test[mask_1],
        mode="markers",
        name="Actual Class 1",
        marker=dict(
            size=6
        )
    )
)



fig.add_trace(
    go.Surface(
        x=X1,
        y=X2,
        z=Z,
        name="Decision Boundary",
        opacity=0.4,
        showscale=False
    )
)




fig.add_trace(
    go.Scatter3d(
        x=X_test[:, 0],
        y=X_test[:, 1],
        z=model._predict_probab(X_test),
        mode="markers",
        name="Predicted Probab",
        marker=dict(
            size=4,
            symbol="diamond"
        )
    )
)



fig.update_layout(

    title="Custom XGBoost Binary Classification",

    scene=dict(
        xaxis_title="Feature 1",
        yaxis_title="Feature 2",
        zaxis_title="Class"
    ),

    width=1000,
    height=750
)


fig.show()
```

```python
xgmodel = XGBClassifier(
    n_estimators=50,
    learning_rate=0.3,
    max_depth=3,
    min_child_weight=1,
    gamma=0,
    reg_lambda=1,
    reg_alpha=0,
    max_leaves=None,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42
)

xgmodel.fit(X_train, y_train)
```

```python
xgpred = xgmodel.predict(X_test)

xg_acc = accuracy_score(y_test, xgpred)

print("Accuracy : ", xg_acc)
```

```python
(xgpred != custom_pred).sum()
```

```python
different = np.where(xgpred != custom_pred)[0]

print("Different indices:", different)

for i in different:
    print(
        "Index:", i,
        "| Actual:", y_test[i],
        "| Custom:", custom_pred[i],
        "| XGBoost:", xgpred[i]
    )
```

```python
custom_proba = model._predict_probab(X_test)
xgb_proba = xgmodel.predict_proba(X_test)[:, 1]

for i in [62, 76]:
    print(
        "Index:", i,
        "| Actual:", y_test[i],
        "| Custom probability:", custom_proba[i],
        "| XGBoost probability:", xgb_proba[i]
    )
```
