```python
import numpy as np 
import pandas as pd 
from Stacking import CustomStackingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split 
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import StackingClassifier
import warnings 

warnings.filterwarnings('ignore')
```

```python
X, y = make_classification(
    n_samples=10000, 
    n_features=3, 
    n_informative=3,
    n_redundant=0,    
    n_classes=3, 
    random_state=42
)

X = pd.DataFrame(X, columns = [f"Feature {i}" for i in [1,2,3]])
```

```python
X
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

scalar = StandardScaler()

X_train = scalar.fit_transform(X_train)
X_test = scalar.transform(X_test)
```

```python
estimators = [LogisticRegression(), KNeighborsClassifier(n_neighbors=7), DecisionTreeClassifier()]

stack = CustomStackingClassifier(estimators=estimators, meta_estimator=LogisticRegression())
stack.fit(X_train, y_train)
```

```python
custom_pred = stack.predict(X_test)
```

```python
acc = accuracy_score(y_test, custom_pred)
print("accuracy : ", acc)
```

```python
from sklearn.metrics import f1_score

f1 = f1_score(y_test, custom_pred, average="macro")
print("f1 : ", f1)
```

```python
estimators = [("LinearRgression" , LogisticRegression()), ("KNN",KNeighborsClassifier(n_neighbors=7)), ("DT", DecisionTreeClassifier())]
```

```python
sk_stack = StackingClassifier(estimators=estimators, final_estimator=LogisticRegression())
sk_stack.fit(X_train, y_train)
```

```python
sk_pred = sk_stack.predict(X_test)

sk_acc = accuracy_score(y_test, sk_pred)

print("Accuracy : ", sk_acc)

sk_f1 = f1_score(y_test, sk_pred, average="macro")

print("sk_rmse : ", sk_f1)
```

```python
pd.DataFrame({
    'custom' : [acc,f1],
    'sklearn': [sk_acc,sk_f1]
}, index=['Accuracy', 'F1_score'])
```

## Conclusion

I implemented a Stacking Classifier from scratch using multiple base classifiers and a Logistic Regression meta-model. I compared it with sklearn's `StackingClassifier` using the same setup. My custom model achieved an accuracy of **0.927576** and an F1-score of **0.927704**, while sklearn achieved an accuracy of **0.930000** and an F1-score of **0.930133**. Both models produced almost identical results, showing that my scratch implementation correctly captures the main working of Stacking Classification.
